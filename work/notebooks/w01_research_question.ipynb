```python
import pandas as pd
import numpy as np

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

total_rows = len(df)
declining_rate = (df["trend_direction"].str.lower() == "down").mean()

age_by_trend = df.groupby("trend_direction")["content_age_days"].median()

print(f"Metric 1 — Total Dataset Size: {total_rows:,} rows")
print(f"Metric 2 — Overall Decline Rate: {declining_rate:.1%}")
print("\nMetric 3 — Median Content Age (days) by Trend Direction:")
print(age_by_trend.to_string())
```

```python
corr = df[['content_age_days', 'days_since_last_update', 'impressions_90d', 'word_count']].corr()
print("Feature Correlation Matrix (Observational Signals Only):")
print(corr.round(3))
```
